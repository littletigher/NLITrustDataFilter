```python
import json
import random

# 读取 generate.json 文件
with open('D:\PycharmProjects\TDFilter\data\generateData\correct\generate.json', 'r') as f:
    generate_data = json.load(f)

# 读取 generate_incorrect.json 文件
with open('D:\PycharmProjects\TDFilter\data\generateData\incorrect\generate_incorrect.json', 'r') as f:
    generate_incorrect_data = json.load(f)

# 计算10%的数据量
sample_size = int(0.9 * len(generate_incorrect_data))

# 从 generate_incorrect 中随机抽取10%的数据
sampled_incorrect_data = random.sample(generate_incorrect_data, sample_size)

# 将抽取的数据与 generate.json 中的数据混合
merged_data = generate_data + sampled_incorrect_data

# 随机打乱混合后的数据
random.shuffle(merged_data)

# 输出到新的文件，假设文件名为 merged_output.json
with open('experiment_data.json', 'w') as f:
    json.dump(merged_data, f, indent=4)

print("数据已成功混合并输出到 experiment_data.json")
```

```python
import json
import pandas as pd
import matplotlib.pyplot as plt

# 假设你的数据是存储在 merged_output.json 文件中的
file_path ='D:/PycharmProjects/TDFilter/experiment4/biologic_experiment.json'

# 读取 JSON 文件
with open(file_path, 'r',encoding='utf-8') as f:
    data = json.load(f)

# 将数据转换为 DataFrame 方便统计
df = pd.DataFrame(data)

# 统计 flag 和 topic 的分布情况
flag_counts = df['flag'].value_counts()
topic_counts = df['topic'].value_counts()

# 绘制 flag 分布的柱状图并保存为独立的图片
fig1, ax1 = plt.subplots(figsize=(8, 6))
ax1.bar(flag_counts.index, flag_counts.values, color='blue')
for i, value in enumerate(flag_counts):
    ax1.text(flag_counts.index[i], value, f'{value}', ha='center', va='bottom')
ax1.set_title('Flag Distribution')
ax1.set_xlabel('Flag')
ax1.set_ylabel('Count')
plt.tight_layout()
plt.savefig('D:/PycharmProjects/TDFilter/experiment4/flag_distribution_with_counts.png')  # 保存 flag 图
plt.close(fig1)

# 绘制 topic 分布的柱状图并保存为独立的图片
fig2, ax2 = plt.subplots(figsize=(10, 8))
ax2.bar(topic_counts.index, topic_counts.values, color='green')
for i, value in enumerate(topic_counts):
    ax2.text(topic_counts.index[i], value, f'{value}', ha='center', va='bottom')
ax2.set_title('Topic Distribution')
ax2.set_xlabel('Topic')
ax2.set_ylabel('Count')
ax2.tick_params(axis='x', rotation=90)
plt.tight_layout()
plt.savefig('D:/PycharmProjects/TDFilter/experiment4/topic_distribution_with_counts.png')  # 保存 topic 图
plt.close(fig2)

# 提示保存成功
print("图像已成功保存为 flag_distribution_with_counts.png 和 topic_distribution_with_counts.png")


```

```python
import json
import random

# 加载原始数据
with open('D:/PycharmProjects/TDFilter/experiment2/experiment_data.json', 'r') as file:
    data = json.load(file)

# 筛选出flag为1的数据
flagged_data = [item for item in data if item['flag'] == 1]

# 提取10%的数据，至少保证提取1条
sample_size = max(1, int(0.1 * len(flagged_data)))
sampled_data = random.sample(flagged_data, sample_size)

# 将提取的数据保存到新的JSON文件中
with open('prestored_data.json', 'w') as outfile:
    json.dump(sampled_data, outfile, indent=4)

print("提取完成，已保存到prestored_data.json文件中。")
```

```python
import json

# 加载两个 JSON 文件
with open('D:\PycharmProjects\TDFilter\experiment3\experiment_data.json', 'r', encoding='utf-8') as f1, open('D:\PycharmProjects\TDFilter\experiment3\prestored_data.json', 'r', encoding='utf-8') as f2:
    data1 = json.load(f1)
    data2 = json.load(f2)

# 找出不同的部分
diff_data1 = [item for item in data1 if item not in data2]
diff_data2 = [item for item in data2 if item not in data1]

# 打印出不同的部分
print("File1 unique entries:")
print(json.dumps(diff_data1, indent=4, ensure_ascii=False))

print("\nFile2 unique entries:")
print(json.dumps(diff_data2, indent=4, ensure_ascii=False))

# 如果你想把差异结果保存到文件中，可以如下保存
with open('diff_file1.json', 'w', encoding='utf-8') as out1:
    json.dump(diff_data1, out1, indent=4, ensure_ascii=False)

with open('diff_file2.json', 'w', encoding='utf-8') as out2:
    json.dump(diff_data2, out2, indent=4, ensure_ascii=False)
```
