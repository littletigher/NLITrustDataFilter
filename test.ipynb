```python
import json

# 给定的 JSON 字符串
json_string = '''{\"conflict_score\": 1,\"explanation\"：\"The validation information states that the Ojai Valley Inn & Spa in California is a luxurious location frequented by celebrities, which aligns with the known information provided.\"}'''
corrected_string = json_string.replace('：', ':')
# 将 JSON 字符串解析为字典
data = json.loads(corrected_string)

# 提取 conflict_score 和 explanation
conflict_score = data['conflict_score']
explanation = data['explanation']
```

```python
from sentence_transformers import SentenceTransformer, util

# 加载预训练的Sentence-BERT模型
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')

# 输入的两个句子
sentence1 = "This is an example sentence."
sentence2 = "This is a similar sentence."

# 计算句子嵌入
embedding1 = model.encode(sentence1, convert_to_tensor=True)
embedding2 = model.encode(sentence2, convert_to_tensor=True)

# 计算余弦相似度
similarity = util.pytorch_cos_sim(embedding1, embedding2)

print(f"Sentence similarity: {similarity.item():.4f}")
```
