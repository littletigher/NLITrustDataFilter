```python
from datasets import load_dataset

# 加载 facebook/asset 数据集
dataset = load_dataset("facebook/asset")

# 查看数据集的信息
print("数据集的键：", dataset.keys())
print("训练集的第一条数据：", dataset["train"][0])

# 遍历数据示例
for i, data in enumerate(dataset["train"][:5]):  # 仅展示前5条数据
    print(f"示例 {i+1}:")
    print(data)
    print("-" * 50)
```

```python
import pandas as pd

# 加载 Parquet 文件
file_path = "validation-00000-of-00001.parquet"
data = pd.read_parquet(file_path)

# 转换为 JSON 格式并保存
json_output_path = "output.json"
data.to_json(json_output_path, orient="records", force_ascii=False)

print(f"Parquet 文件已成功转换为 JSON 格式，保存为：{json_output_path}")
```
