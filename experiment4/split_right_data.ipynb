```python
import json

# 假设您的 JSON 文件名为 data.json
with open('experiment_data.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

# 检查数据类型是否为列表
if isinstance(data, list):
    # 过滤出 flag == 1 的数据
    filtered_data = [item for item in data if item.get('flag') == 1]
else:
    print("JSON 文件的顶层结构不是列表，请检查文件格式。")
    filtered_data = []

# 将结果保存到新的 JSON 文件中
with open('filtered_flag_1_data.json', 'w', encoding='utf-8') as f:
    json.dump(filtered_data, f, ensure_ascii=False, indent=4)

print("已提取 flag=1 的数据，并保存到 filtered_flag_1_data.json 文件中。")
```
