```python
import json

data = []
file_path = 'radiologicalError.json'

with open(file_path, 'r', encoding='utf-8') as file:
    for line in file:
        line = line.strip(',\n')  # 去除逗号和换行符
        json_data = json.loads(line)  # 正确调用 json.loads
        if "generateKnowledge" in json_data and "basicKnowledge" in json_data and "flag" in json_data:
            data.append({
                "generateKnowledge": json_data["generateKnowledge"],
                "basicKnowledge": json_data["basicKnowledge"],
                "flag": json_data["flag"]
            })

len(data)
```
