```python
import json

text_data = []
# 2. 从 JSON 文件加载测试数据
with open(r'D:\PycharmProjects\TDFilter\experiment5\gpt35\Biological\iterate\results.jsonl', 'r',encoding='utf-8') as f:
    for line in f:
        text_data.append(json.loads(line))
```

```python
from sentence_transformers import SentenceTransformer, util
import numpy as np

# 示例数据
data = text_data

# 提取知识点
knowledge_texts = [item['knowledge'] for item in data]

# 加载预训练的 Sentence-BERT 模型
model = SentenceTransformer('all-MiniLM-L6-v2')  # 或 'all-MPNet-base-v2' 等模型

# 生成嵌入
embeddings = model.encode(knowledge_texts, convert_to_tensor=True)

# 计算相似性矩阵
cosine_similarities = util.cos_sim(embeddings, embeddings).cpu().numpy()

# 提取矩阵的上三角（不包括对角线）部分
num_items = len(knowledge_texts)
upper_triangle = [
    cosine_similarities[i][j]
    for i in range(num_items) for j in range(i + 1, num_items)
]

# 计算平均相似性
average_similarity = np.mean(upper_triangle)

# 输出结果
print(f"知识点的平均相似性: {average_similarity:.4f}")
```
